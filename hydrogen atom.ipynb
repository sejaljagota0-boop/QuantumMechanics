```python
import numpy as np
import matplotlib.pyplot as plt

a0 = 1.0                      
r = np.linspace(0, 20*a0, 500)


R10 = 2/a0**1.5 * np.exp(-r/a0)
R20 = 1/(np.sqrt(2)*a0**1.5) * (1 - r/(2*a0)) * np.exp(-r/(2*a0))
R21 = 1/(2*np.sqrt(6)*a0**1.5) * (r/a0) * np.exp(-r/(2*a0))


plt.figure(figsize=(7, 4.5))
plt.plot(r/a0, R10, label="R$_{10}$ (1s)")
plt.plot(r/a0, R20, label="R$_{20}$ (2s)")
plt.plot(r/a0, R21, label="R$_{21}$ (2p)")
plt.axhline(0, color="gray", lw=0.5)
plt.xlabel("r / a$_0$")
plt.ylabel("R(r)")
plt.title("Radial functions of the hydrogen atom")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("hydrogen_simple.png", dpi=200)
plt.show()
```
